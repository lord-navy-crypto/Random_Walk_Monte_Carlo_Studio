```
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# 1. Basic 2D plots: curves, scatter points, and error bars
x = np.linspace(0, 2*np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y_err = 0.1 * np.random.rand(len(x))  # Random error values

plt.figure(figsize=(10, 4))
# Panel 1: curve and scatter plot
plt.subplot(121)
plt.plot(x, y1, 'r-', label='sin(x)', linewidth=2)  # Red curve
plt.scatter(x[::10], y2[::10], c='blue', label='cos(x)')  # Every tenth point
plt.xlabel('x (rad)')
plt.ylabel('y')
plt.title('Sin/Cos Curve + Scatter')
plt.legend()
plt.grid(True)

# Panel 2: error-bar plot
plt.subplot(122)
plt.errorbar(x[::5], y1[::5], yerr=y_err[::5], fmt='o-', ecolor='gray', capsize=3)
plt.xlabel('x (rad)')
plt.ylabel('sin(x) ± error')
plt.title('Error Bar Plot')
plt.tight_layout()  # Adjust the layout automatically
plt.savefig('matplotlib_2d_test.png', dpi=300)  # Save a high-resolution image

# 2. 3D surface plot
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
X = np.linspace(-5, 5, 50)
Y = np.linspace(-5, 5, 50)
X, Y = np.meshgrid(X, Y)
Z = np.sin(np.sqrt(X**2 + Y**2))  # Three-dimensional surface

ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('3D Surface Plot (sin(√(X²+Y²)))')
plt.savefig('matplotlib_3d_test.png', dpi=300)
plt.show()  # Display all figures
```
